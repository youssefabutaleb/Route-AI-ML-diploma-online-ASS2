```python
def ret_str(ls,k):
    for s in ls:
        if len(s)==k:
            ls.remove(s)
    return ls

y = ret_str(['apple', 'pie', 'banana'], 3)
print(y)
```

```python
import math

def area(shape,r,l=None):
    if shape.lower() == 'circle':
        return math.pi * (r**2)
    elif shape.lower() == 'rectangle':
        return l*r
    
c = round(area("circle", 5),2)
r=round(area("rectangle", 4, 6),2)
print(c)
print(r)
```

```python
def sum_(l):
    #list of integers
    ints = [str(i) for i in range(1000)]
    
    sum1=0
    for x in l:
        if type(x) == int:
            sum1+=x
        elif x in ints: #if the users enters an int in the form of str like "20"
            sum1+=int(x)
        else:
            continue
    return sum1

ls = [10, 'apple', 20, True, 5]
s = sum_(ls)

print(s)

string_ints = ["10", "20", "30"]
si = sum_(string_ints)
print(si)
```

```python
class Restaurant:
    #initilization
    def __init__(self, menu_items={}, book_table=[], customer_orders=[]):
        self.menu_items = menu_items
        self.book_table = book_table
        self.customer_orders = customer_orders
    #adding
    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price
        
    #reservatiom
    def book_tables(self, table_number):
        if table_number not in self.book_table:
            self.book_table.append(table_number)
            
    #orders
    def take_order(self, table_number, order_items):
        order={
            "table": table_number,
            "items": order_items
        }
        self.customer_orders.append(order)
        
    #menu
    def disp_menu(self):
        for k,v in self.menu_items.items():
            print(f"{k}: ${v}]")
            
    #reserved tables
    def disp_reserved(self):
        for table in self.book_table:
            print(table)    
            
    #display order
    def disp_orders(self):
        for order in self.customer_orders:
            print(f"Table: {order['table']} | Items: {', '.join(order['items'])}")
```

Question 5 tasks

```python
my_restaurant = Restaurant()

# Task 1: Add items to the menu
my_restaurant.add_item_to_menu("Burger", 10.99)
my_restaurant.add_item_to_menu("Pizza", 12.50)
my_restaurant.add_item_to_menu("Salad", 7.00)

# Task 2: Make table reservations
my_restaurant.book_tables("Table 1")
my_restaurant.book_tables("Table 3")

# Task 3: Take customer orders
my_restaurant.take_order("Table 1", ["Burger", "Salad"])
my_restaurant.take_order("Table 3", ["Pizza"])

# Task 4: Print the menu
my_restaurant.disp_menu()

# Task 5: Print table reservations
my_restaurant.disp_reserved()

# Task 6: Print customer orders
my_restaurant.disp_orders()
```
